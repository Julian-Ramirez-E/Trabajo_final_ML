# Cargar_datos.ipynb
**Objetivo:** leer el archivo Excel original (`BD_creditos (2).xlsx`) y guardarlo como `Base_de_datos.csv` en la raíz del repositorio.

> En una empresa real este paso no existiría: los datos vendrían de un Data Warehouse o Data Lake. Aquí usamos un archivo de ejemplo.

```python
# 1. Importar la librería pandas (sirve para trabajar con tablas)
import pandas as pd
```

```python
# 2. Definir las rutas de los archivos
# El notebook está dentro de mlops_pipeline/, por eso usamos ".." para subir
# una carpeta y llegar a la raíz del repositorio.
ruta_excel = "../BD_creditos (2).xlsx"   # archivo de entrada
ruta_csv = "../Base_de_datos.csv"        # archivo de salida
```

```python
# 3. Leer el Excel y guardarlo en una tabla llamada df (DataFrame)
# Necesita la librería openpyxl (ya está en requirements.txt)
df = pd.read_excel(ruta_excel)

print("Filas y columnas:", df.shape)
```

```python
# 4. Ver las primeras 5 filas para confirmar que se leyó bien
df.head()
```

```python
# 5. Ver los tipos de dato de cada columna
# (más adelante, en el EDA, corregiremos los que estén mal)
df.info()
```

```python
# 6. Guardar la tabla como CSV
# index=False evita que pandas agregue una columna extra con el número de fila
df.to_csv(ruta_csv, index=False)

print("Archivo guardado en:", ruta_csv)
```

```python
# 7. Comprobación: leer el CSV recién creado y verificar que tiene el mismo tamaño
df_check = pd.read_csv(ruta_csv)
print("Tamaño del Excel:", df.shape)
print("Tamaño del CSV  :", df_check.shape)
```
